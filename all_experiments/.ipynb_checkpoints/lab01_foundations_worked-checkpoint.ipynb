## Part B — Titanic data quality check

Look at the data before touching a model: how big it is, what each column holds, what is missing, and how balanced the label is.

```python
df = pd.read_csv(DATA / "titanic_dataset.csv")

print("shape:", df.shape)
print(df.head())
```

```python
# Four questions to answer before modelling: what types, what is missing,
# how balanced is the label, and which columns are identifiers?
print("Data types:\n", df.dtypes)
print("\nMissing values:\n", df.isna().sum().sort_values(ascending=False))
print("\nSurvival rate:\n", df["Survived"].value_counts(normalize=True))
```

```python
# Two simple features that later labs reuse. Both use only information
# that would be known before the outcome.
df["FamilySize"] = df["SibSp"] + df["Parch"] + 1
df["IsAlone"] = (df["FamilySize"] == 1).astype(int)

summary = (
    df.groupby(["Sex", "Pclass"])["Survived"]
      .agg(["count", "mean"])
      .rename(columns={"mean": "survival_rate"})
)
print("\nSurvival by sex and class:\n", summary)
```

**What the output shows.** 891 passengers, 12 columns. `Cabin` is missing for 77% of rows, `Age` for 20%, `Embarked` for 2 rows. Only 38% survived, so a model that always says *did not survive* is already 62% accurate; anything we build must beat that. Survival ranges from 97% (first-class women) to 14% (third-class men), so `Sex` and `Pclass` will matter. `PassengerId`, `Name`, `Ticket`, and `Cabin` identify rows rather than describe passengers, so they are not features.

## Part C — The same checklist on an unfamiliar table

Now a table nobody has seen before. The same checklist applies, plus one question Chapter 3 insists on: *when* would the prediction be made, and which columns exist at that moment?

```python
# Student Performance: 649 Portuguese secondary-school students, 33 columns.
# Nobody in the room knows this table yet, so the checklist has to do the work.
students = pd.read_csv(DATA / "student_performance_dataset.csv")

print("shape:", students.shape)
print("duplicate rows:", students.duplicated().sum())
print("missing values:", students.isna().sum().sum())
print("\nColumn types:\n", students.dtypes.value_counts())
print("\nText columns:", students.select_dtypes(exclude="number").columns.tolist())
```

```python
# The label is the final grade G3 (0-20). G1 and G2 are the two earlier term grades.
print(students["G3"].describe().round(1))
print("\nStudents with G3 = 0:", int((students["G3"] == 0).sum()))
print("\nCorrelation with G3:\n",
      students[["G1", "G2", "absences", "studytime", "failures"]].corrwith(students["G3"]).round(2))
```

```python
# The prediction-moment test. When would the prediction be used?
#   - at enrolment: G1 and G2 do not exist yet, so they cannot be features;
#   - after the second term: G1 and G2 are known and are the strongest features.
# Both are valid problems, but they are different problems with different features.
at_enrolment = [c for c in students.columns if c not in {"G1", "G2", "G3"}]
after_term_two = [c for c in students.columns if c != "G3"]
print("features available at enrolment:", len(at_enrolment))
print("features available after term two:", len(after_term_two))
```

**What the output shows.** 649 students, 33 columns, no missing values and no duplicate rows: a clean table, which is itself worth recording. Seventeen columns are text and will need encoding. The final grade `G3` runs from 0 to 20 with a median of 12; fifteen students scored 0, which looks like a code for *did not sit the exam* rather than a real grade and should be checked. `G2` correlates 0.92 with `G3` and `G1` 0.83, while `absences` barely correlates at all. That is the prediction-moment test in numbers: if the prediction is made at enrolment, the two strongest columns do not exist yet and the problem is much harder (30 features); after term two they exist and the problem is easy (32 features). Same file, two different projects.

## Evidence

Write the numbers above to the results folder. These files are regenerated on every run; do not edit them by hand.
