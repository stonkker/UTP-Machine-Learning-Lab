## Setup

Imports and paths. Notebooks run from `all_experiments/`, so the course folder is one level up; figures and tables go to `results/lab03/`.

```python
import json
from pathlib import Path

import joblib
import matplotlib.pyplot as plt
import pandas as pd
import sklearn
from sklearn.compose import ColumnTransformer
from sklearn.ensemble import RandomForestClassifier
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    ConfusionMatrixDisplay,
    PrecisionRecallDisplay,
    RocCurveDisplay,
    accuracy_score,
    average_precision_score,
    classification_report,
    confusion_matrix,
    roc_auc_score,
)
from sklearn.model_selection import GridSearchCV, StratifiedKFold, train_test_split
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler

PROJECT_ROOT = Path.cwd().parent
DATA = PROJECT_ROOT / "all_datasets"
MODELS = PROJECT_ROOT / "models"
MODELS.mkdir(exist_ok=True)

# One helper builds the same preprocessing for any table: numbers are
# imputed and scaled, text columns are imputed and one-hot encoded.
def make_preprocess(num_cols, cat_cols):
    return ColumnTransformer([
        ("num", Pipeline([
            ("impute", SimpleImputer(strategy="median")),
            ("scale", StandardScaler()),
        ]), num_cols),
        ("cat", Pipeline([
            ("impute", SimpleImputer(strategy="most_frequent")),
            ("onehot", OneHotEncoder(handle_unknown="ignore")),
        ]), cat_cols),
    ])

RESULTS = PROJECT_ROOT / "results" / "lab03"
RESULTS.mkdir(parents=True, exist_ok=True)
print("course folder:", PROJECT_ROOT.name)
```

## Part A — Two baselines on Titanic

Two models, one pipeline, one split. Logistic regression is the simple reference; the random forest is the more flexible challenger.

```python
titanic = pd.read_csv(DATA / "titanic_dataset.csv")

features = ["Pclass", "Sex", "Age", "SibSp", "Parch", "Fare", "Embarked"]
num_cols = ["Age", "SibSp", "Parch", "Fare"]
cat_cols = ["Pclass", "Sex", "Embarked"]

X = titanic[features]
y = titanic["Survived"]

X_train, X_valid, y_train, y_valid = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)
```

```python
candidates = {
    "logistic": LogisticRegression(max_iter=1000),
    "random_forest": RandomForestClassifier(
        n_estimators=300, min_samples_leaf=2, random_state=42
    ),
}

for name, estimator in candidates.items():
    pipe = Pipeline([
        ("preprocess", make_preprocess(num_cols, cat_cols)),
        ("model", estimator),
    ])
    pipe.fit(X_train, y_train)
    pred = pipe.predict(X_valid)
    prob = pipe.predict_proba(X_valid)[:, 1]
    print(f"{name:14s} accuracy={accuracy_score(y_valid, pred):.3f} "
          f"roc_auc={roc_auc_score(y_valid, prob):.3f}")
```

**What the output shows.** Logistic: accuracy 0.776, ROC AUC 0.842. Random forest: accuracy 0.798, ROC AUC 0.834. The two metrics disagree about the winner, and on 223 validation rows neither gap is large. Part D settles it with cross-validation.

## Part B — Thresholds and ranking metrics

A classifier outputs a probability. Turning it into a yes/no decision needs a threshold, and that threshold is a choice. See what moving it from 0.50 to 0.30 does, then look at two metrics that do not depend on it.

```python
cancer = pd.read_csv(DATA / "breast_cancer_dataset.csv")

X_c = cancer.drop(columns=["id", "diagnosis", "Unnamed: 32"], errors="ignore")
y_c = (cancer["diagnosis"] == "M").astype(int)

Xc_train, Xc_valid, yc_train, yc_valid = train_test_split(
    X_c, y_c, test_size=0.25, random_state=42, stratify=y_c
)

cancer_model = Pipeline([
    ("scale", StandardScaler()),
    ("model", LogisticRegression(max_iter=2000)),
])
cancer_model.fit(Xc_train, yc_train)
prob = cancer_model.predict_proba(Xc_valid)[:, 1]
```

```python
# The model gives a probability. The threshold that turns it into a
# decision is a choice, and we make that choice on validation data.
for threshold in [0.50, 0.30]:
    pred = (prob >= threshold).astype(int)
    print(f"\nthreshold = {threshold:.2f}")
    print(confusion_matrix(yc_valid, pred))
    print(classification_report(yc_valid, pred, digits=3))
    ConfusionMatrixDisplay.from_predictions(yc_valid, pred)
    plt.title(f"Confusion matrix at threshold {threshold:.2f}")
    plt.tight_layout()
    plt.savefig(RESULTS / f"confusion_matrix_at_threshold_{threshold:.2f}.png", dpi=160, bbox_inches="tight")
    plt.show()
```

```python
# These two metrics summarise the ranking and do not depend on a threshold.
print("ROC AUC:", round(roc_auc_score(yc_valid, prob), 3))
print("PR AUC:", round(average_precision_score(yc_valid, prob), 3))

RocCurveDisplay.from_predictions(yc_valid, prob)
plt.title("ROC curve")
plt.tight_layout()
plt.savefig(RESULTS / "roc_curve.png", dpi=160, bbox_inches="tight")
plt.show()

PrecisionRecallDisplay.from_predictions(yc_valid, prob)
plt.title("Precision-recall curve")
plt.tight_layout()
plt.savefig(RESULTS / "precision_recall_curve.png", dpi=160, bbox_inches="tight")
plt.show()
```

**What the output shows.** At 0.50 the model misses 4 of 53 malignant cases with 1 false alarm. At 0.30 it misses only 2, still with 1 false alarm. ROC AUC 0.996 and PR AUC 0.994 say the ranking is almost perfect; the remaining errors are about where the cut is, not about the model. The decision memo at the end asks you to defend a threshold.

## Part C — Optional: the same workflow on heart-failure data

Same workflow, different table. The column lists come from the data types, so nothing needs editing by hand.

```python
heart = pd.read_csv(DATA / "heart_failure_dataset.csv")

X_h = heart.drop(columns="HeartDisease")
y_h = heart["HeartDisease"]

# Column lists come from the data types, so the pipeline adapts by itself.
h_num = X_h.select_dtypes(include="number").columns.tolist()
h_cat = X_h.select_dtypes(exclude="number").columns.tolist()

Xh_train, Xh_valid, yh_train, yh_valid = train_test_split(
    X_h, y_h, test_size=0.25, random_state=42, stratify=y_h
)
heart_model = Pipeline([
    ("preprocess", make_preprocess(h_num, h_cat)),
    ("model", LogisticRegression(max_iter=2000)),
])
heart_model.fit(Xh_train, yh_train)
prob_h = heart_model.predict_proba(Xh_valid)[:, 1]
pred_h = (prob_h >= 0.50).astype(int)

print(classification_report(yh_valid, pred_h, digits=3))
print("ROC AUC:", round(roc_auc_score(yh_valid, prob_h), 3))
print("PR AUC:", round(average_precision_score(yh_valid, prob_h), 3))

ConfusionMatrixDisplay.from_predictions(yh_valid, pred_h)
plt.title("Heart-disease confusion matrix")
plt.tight_layout()
plt.savefig(RESULTS / "heart_disease_confusion_matrix.png", dpi=160, bbox_inches="tight")
plt.show()
```

**What the output shows.** Accuracy 0.883, ROC AUC 0.932, recall 0.921 for heart disease at the default threshold. Only the target column name changed; the pipeline adapted itself to five text and six numeric columns.

## Part D — Tune with cross-validation, then test once

Tune on training data with cross-validation, lock the choice, then open the test set once.

```python
# A fresh split: X_train is for all development, X_test is opened once at the end.
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

pipe = Pipeline([
    ("preprocess", make_preprocess(num_cols, cat_cols)),
    ("model", LogisticRegression(max_iter=2000)),
])

search = GridSearchCV(
    pipe,
    param_grid={
        "model__C": [0.01, 0.1, 1.0, 10.0, 100.0],
        "model__class_weight": [None, "balanced"],
    },
    scoring="roc_auc",
    cv=StratifiedKFold(n_splits=5, shuffle=True, random_state=42),
    n_jobs=-1,
)
search.fit(X_train, y_train)
```

```python
# The choice is locked. Now, and only now, evaluate on the test set.
test_prob = search.predict_proba(X_test)[:, 1]
test_auc = roc_auc_score(y_test, test_prob)

print("best parameters:", search.best_params_)
print("cross-validated ROC AUC:", round(search.best_score_, 3))
print("final test ROC AUC:", round(test_auc, 3))
```

**What the output shows.** The search picks `C = 0.1` with balanced class weights: cross-validated ROC AUC 0.850, then a single test ROC AUC of 0.837. The test score is a little lower, as expected: the CV score was used to pick the best of ten options, so it is slightly optimistic. Report the test number once and do not tune again after seeing it.

## Part E — Save the final model and use it

Save the whole pipeline to a file, write a small model card, load the file back, and predict for two passengers who were never in the data. This is what deploying a model means.

```python
# The whole pipeline is the model: preprocessing and classifier together.
final_model = search.best_estimator_
model_path = MODELS / "titanic_survival.joblib"
joblib.dump(final_model, model_path)

# A model card records what the file expects and how well it did.
card = {
    "task": "predict Titanic survival (1 = survived)",
    "features": features,
    "model": "logistic regression pipeline",
    "best_parameters": search.best_params_,
    "training_rows": int(len(X_train)),
    "final_test_roc_auc": round(float(test_auc), 3),
    "decision_threshold": 0.5,
    "scikit_learn_version": sklearn.__version__,
}
(MODELS / "titanic_survival_card.json").write_text(json.dumps(card, indent=2))
print("saved:", model_path.name, "and titanic_survival_card.json")
```

```python
# Deployment in one line: load the file, give it new rows with the same
# column names, and read the predictions. No retraining, no preprocessing code.
loaded = joblib.load(model_path)

new_passengers = pd.DataFrame([
    {"Pclass": 1, "Sex": "female", "Age": 29, "SibSp": 0, "Parch": 0, "Fare": 80.0, "Embarked": "C"},
    {"Pclass": 3, "Sex": "male",   "Age": 40, "SibSp": 0, "Parch": 0, "Fare":  7.9, "Embarked": "S"},
])
new_passengers["survival_probability"] = loaded.predict_proba(new_passengers)[:, 1].round(3)
print(new_passengers)
```

**What the output shows.** `models/titanic_survival.joblib` is about 5 KB and contains the imputers, the scaler, the encoder, and the classifier. The reloaded model gives a first-class woman a 94% chance of survival and a third-class man 14%. The card records what the file expects (the seven feature names), the threshold, the test score, and the library version. `predict_passenger.py` in this folder uses the same two files from the terminal.
